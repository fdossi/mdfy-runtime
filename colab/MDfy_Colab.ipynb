## 2. Enviar e converter
Execute, selecione um ou mais arquivos e aguarde os downloads em Markdown.

```python
from google.colab import files
from runtime.converter import convert_upload

# Selecione os arquivos no seu dispositivo.
uploaded=files.upload()

# Converte e baixa cada resultado em formato .md.
for name,content in uploaded.items():
    result=convert_upload({'name':name,'content':content})
    path='/content/'+result['name']
    open(path,'wb').write(result['content'])
    files.download(path)
```
