```python
from google.colab import files
from runtime.converter import convert_upload
uploaded=files.upload()
for name,content in uploaded.items():
    result=convert_upload({'name':name,'content':content})
    path='/content/'+result['name']
    open(path,'wb').write(result['content'])
    files.download(path)
```
