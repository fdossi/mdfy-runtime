# MDfy — Conversor avançado
Envie novamente o arquivo que exige processamento em servidor. A sessão é temporária; baixe o resultado antes de fechar esta página.

```python
from ipywidgets import FileUpload, Button, HTML, Output, VBox
from IPython.display import display, FileLink
from runtime.converter import convert_upload
from pathlib import Path

upload=FileUpload(accept='.pdf,.doc,.docx,.xls,.xlsx,.csv,.epub,.mobi,.png,.jpg,.jpeg,.tif,.tiff,.djvu,.zip,.tar,.tgz',multiple=False)
button=Button(description='Converter para Markdown',button_style='success',icon='cogs')
status=HTML(value='<b>Selecione um arquivo de até 100 MB.</b>')
output=Output()

def convert(_):
    output.clear_output(); status.value='<b>Convertendo…</b>'
    if not upload.value: status.value='<span style="color:#a33">Selecione um arquivo.</span>'; return
    item=upload.value[0]
    if item['size']>100*1024*1024: status.value='<span style="color:#a33">O limite é 100 MB.</span>'; return
    try:
        result=convert_upload({'name':item['name'],'content':bytes(item['content'])})
        folder=Path('outputs'); folder.mkdir(exist_ok=True)
        target=folder/result['name']; target.write_bytes(result['content'])
        status.value='<b style="color:#08754f">Conversão concluída.</b>'
        with output: display(FileLink(target,result_html_prefix='Baixar: '))
    except Exception as error: status.value=f'<span style="color:#a33">Falha: {error}</span>'
button.on_click(convert)
display(VBox([HTML('<h2 style="color:#0c8058">MDfy avançado</h2>'),upload,button,status,output]))
```
